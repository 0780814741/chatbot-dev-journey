# Pizza Order Bot
Exercise 1 - Week 1 Activity

```python
print("Hi! Welcome to the Pizza Order Bot.")
print("Type 'bye' anytime to quit.")

quit_now = False

while True:
    size = input("What size pizza do you want? (Small, Medium, Large): ")
    if size.lower() == "bye":
        quit_now = True
        break
    if size.lower() in ["small", "medium", "large"]:
        break
    print("Sorry, I don't understand. Please choose Small, Medium or Large.")

if quit_now:
    print("Goodbye!")
else:
    while True:
        topping = input("What topping do you want? (Pepperoni, Mushrooms, Cheese): ")
        if topping.lower() == "bye":
            quit_now = True
            break
        if topping.lower() in ["pepperoni", "mushrooms", "cheese"]:
            break
        print("Sorry, I don't understand. Please choose Pepperoni, Mushrooms or Cheese.")

    if quit_now:
        print("Goodbye!")
    else:
        print("Awesome, I have recorded a", size.capitalize(), topping.capitalize(), "pizza for you! Goodbye!")
```
